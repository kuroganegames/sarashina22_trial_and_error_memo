```python
from datasets import load_dataset, concatenate_datasets
from datasets import Dataset
from tqdm import tqdm
```

```python
SEED       = 20250715        # 何度実行しても同じになる任意の整数
SAMPLE_SZ  = 100_000
# DIR_CACHE = r"/media/kurogane/HD-NRLD-A/cache"

TARGET_SPLITS = "split_0"
DATASET_ID = "nvidia/OpenCodeReasoning"
DIR_SAVE = "random_code_100k_split_0"
```

```python
ds  = load_dataset(DATASET_ID, TARGET_SPLITS, cache_dir=DIR_CACHE)[TARGET_SPLITS]
```

```python
subset = ds.shuffle(seed=SEED)            # seed を固定:contentReference[oaicite:1]{index=1}
subset = subset.select(range(SAMPLE_SZ))  # 100k 件だけ取得:contentReference[oaicite:2]{index=2}
```

```python
l_convrsatuons = []
for i_subset in tqdm(subset):
    i_data = {
        "id": i_subset["id"],
        "conversations": [
        {
            "role": "user",
            "content": i_subset["input"]
        },
        {
            "role": "assistant",
            "content": i_subset["output"].split("</think>")[1]
        },
    ]
    }
    l_convrsatuons.append(i_data)
```

```python
ds_reconst = Dataset.from_list(l_convrsatuons)
ds_reconst.save_to_disk(DIR_SAVE)
```
