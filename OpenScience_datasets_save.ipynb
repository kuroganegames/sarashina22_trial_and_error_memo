```python
import json
from tqdm import tqdm
import jsonlines
```

```python
dir_jsonl = "OS-Q3-235B-4.jsonl"
```

```python
l_dataset = []
with open(dir_jsonl, 'r', encoding='utf-8') as reader:

    for line in tqdm(reader):
        data = json.loads(line)
        d_temp_dataset = {"conversations": [
            {
            "content": data["input"],
            "role": "user"
            },
            {
            "content": data["output"].split("</think>\n\n")[1],
            "role": "assistant"
            },
        ]}
        l_dataset.append(d_temp_dataset)
```

```python
from datasets import Dataset
```

```python
dataset = Dataset.from_list(l_dataset)
```

```python
dataset.save_to_disk('OpenScience-OS-Q3-235B-4')
```
